```python
import os
import dill
import pathos
from pathlib import Path
from functools import partial
```

```python
def __save_results(
    save_dir,    
    x
):
    # Save results
    with (save_dir/'{}.pkl'.format(x)).open('wb') as f:
        dill.dump([x], f)
```

```python

# Save path
save_dir = Path('./output/test_async_io/')

# Get function for executing
exec_fn = partial(
    __save_results,
    save_dir
)
```

```python
# Run experiment for each realization
iterable = range(50)
with pathos.pools.ProcessPool(os.cpu_count()) as p:
    p.map(
        exec_fn, 
        iterable
    )
```

```python
for i in iterable:
    with (save_dir/'{}.pkl'.format(i)).open('rb') as f:
            print(dill.load(f))
```
