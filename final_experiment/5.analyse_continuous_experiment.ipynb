```python
import os
import sys
import dill
import pprint
from pathlib import Path
import itertools
import pathos
sys.path.append('/Users/leonardo.tessarolo/git/bayesian_bss/')

import math
import numpy as np

from src import LoggingUtilities, VisualizationUtilities, ContinuousExperimentParser
np.random.seed(100)
```

# 1. Configurations

```python
# Whether or not to filter only converged realizations
FILTER_CONVERGED=True

# Number of sources and observations
NSOURCES=2
NOBS=1000

# Mixing matrix configuration
A = np.array([
    [math.sqrt(2)/2, -math.sqrt(2)/2],
    [math.sqrt(2)/2, math.sqrt(2)/2]
])

# Ideal separating matrix
B = np.linalg.inv(A)

# Whether or not to create config
EXECUTE=True

# Whether or not to create folder structure
CREATE_FOLDER_STRUCTURE=True

# Experiment name
EXPERIMENT_NAME='continuous_exp_50'

# Folder which will contain output directory tree
OUTPUT_DIR='./output'
base_output_path=Path(OUTPUT_DIR)

# Create experiment directory
experiment_dir = base_output_path / EXPERIMENT_NAME
LoggingUtilities.configure_logging(
    log_dir=experiment_dir,
    log_type='continuous'
)

# Parse
parser=ContinuousExperimentParser(
    experiment_dir=experiment_dir,
    filter_converged=FILTER_CONVERGED
)
parser.parse(verbose=True)
```

```python
# Initialize folder structure, if so specified
if CREATE_FOLDER_STRUCTURE:
    diagnostics_dir = experiment_dir / 'diagnostics'
    trace_plots_dir = experiment_dir / 'trace_plots'
    average_weights_dir = experiment_dir / 'average_weights'
    estimates_dir = experiment_dir / 'estimates'
    sensitivities_dir = experiment_dir / 'sensitivities'
    
    # Creates dirs for storing outputs, if not yet created
    if not diagnostics_dir.is_dir():
        diagnostics_dir.mkdir()
    if not trace_plots_dir.is_dir():
        trace_plots_dir.mkdir()
    if not average_weights_dir.is_dir():
        average_weights_dir.mkdir()
    if not estimates_dir.is_dir():
        estimates_dir.mkdir()
    if not sensitivities_dir.is_dir():
        sensitivities_dir.mkdir()
```
